Los valores son cosenos continuos; N=3 es un smoke test, no una conclusión de clonación.

```python
# Ejecutar antes las celdas de instalación y preparación de 00_xtts_smoke_test.ipynb en este runtime.
import json, os, sys, subprocess
from pathlib import Path
ROOT=Path('/content/Proyecto-IA-Traductor'); assert ROOT.exists(), 'Clone el repositorio y ejecute primero el notebook XTTS.'
os.chdir(ROOT); sys.path.insert(0,str(ROOT))
from src.evaluation.speaker_similarity import speaker_similarity
from src.voice_cloning.audio import validate_audio
rows=json.loads((ROOT/'outputs/smoke_tests/xtts/results.json').read_text(encoding='utf-8'))
results=[]
for row in rows:
    original=validate_audio(row['source_audio']); generated=validate_audio(row['output_audio'])
    results.append({'sample_id':row['sample_id'],'cosine_similarity':speaker_similarity(row['source_audio'],row['output_audio']),'duration_original':original['duration_sec'],'duration_generated':generated['duration_sec'],'XTTS_time':row['inference_time_sec']})
out=ROOT/'outputs/smoke_tests/voice_similarity.json'; out.write_text(json.dumps(results,ensure_ascii=False,indent=2),encoding='utf-8')
values=[x['cosine_similarity'] for x in results]; print(results); print({'mean':sum(values)/len(values),'min':min(values),'max':max(values),'n':len(values)})
```
